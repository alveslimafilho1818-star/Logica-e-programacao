```

#1- Observe o código apresentado a seguir e veja outras formas de inserir dados em uma lista.

#A) Refaça o código utilizando a estrutura FOR:
numeros = [0, 0, 0, 0, 0]

for x in range(5):
    numeros[x] = float(input("Digite um número qualquer: "))

for y in range(5):
    print(f"{numeros[y]}")

#B) O que acontece se colocarmos while x <= 4? Justifique:
#RESPOSTA: Funcionalmente, terá o mesmo resultado, pois os índices de uma lista de tamanho 5 variam de 0 a 4. As condições x < 5 e x <= 4 executam exatamente para x = 0, 1, 2, 3, 4.

#C) Esse código funcionaria caso quiséssemos inserir mais de 5 elementos? Justifique:
#RESPOSTA: Não, pois como a lista foi inicializada com 5 elementos e o valor do while está fixado em 5. Caso tentarmos acessar um valor maior, acorrerá um erro, pois o índice estará fora do alcance.

#D) Qual a problemática da alternativa anterior (c) para a realidade dos códigos?
#RESPOSTA: Utilizar um tamanho fixo engessa o programa, impedindo que ele trate dados de tamanho dinâmico ou definido pelo usuário em tempo de execução.

```

```
#E) Refaça o código melhorando os problemas encontrados na alternativa (c e d):

numeros = []
qtd = int(input("Quantos números deseja inserir? "))

for i in range(qtd):
    num = float(input(f"Digite o {i+1}º número: "))
    numeros.append(num)

print("\nNúmeros digitados:")
for num in numeros:
    print(num)
```

```
#4. Faça um programa que percorra duas listas, A e B, e gere uma terceira sem elementos repetidos.
A = ['Python', 'Java', 'C', 'PHP', 'JavaScript', 'Dart']
B = ['C++', 'Python', 'Java', 'Julia', 'Go', 'JavaScript']

C = []

# Adiciona elementos de A
for item in A:
    if item not in C:
        C.append(item)

# Adiciona elementos de B
for item in B:
    if item not in C:
        C.append(item)

print("União das listas sem duplicatas:", C)
```

```
#5. Observe o código a seguir e analise as questões pedidas

#A) Explique a função das linhas 20, 24, 25 e 27

#RESPOSTA:
#Linha 20 (achou = False): Inicializa a variável como flag/sinalizadora com valor falso, indicando que o elemento ainda não foi encontrado.

#Linha 24 (achou = True): Altera o valor da flag para verdadeiro assim que o elemento procurado é localizado na lista.

#Linha 25 (break): Interrompe imediatamente a execução do laço while, evitando verificações desnecessárias nos elementos restantes.

#Linha 27 (if achou:): Testa se a busca foi bem-sucedida para decidir qual mensagem exibir na tela

#B) Refaça o código de forma a realizar a mesma tarefa, mas sem utilizar a variável achou:

L = [15, 7, 27, 39]
p = int(input("Digite o valor a procurar: "))

x = 0
while x < len(L):
    if L[x] == p:
        print(f"{p} achado na posição {x}")
        break
    x += 1
else:
    # A cláusula else do while só executa se o laço terminar sem 'break'
    print(f"{p} não encontrado")
```

```
#6. Observe o código a seguir e analise as questões pedidas.

#A) Modifique o código usando a estrutura FOR para as duas repetições:

# O laço principal precisa ser um while True para ler entradas até o usuário digitar 0.
# O laço interno pode usar FOR para imprimir a lista.

L = []
while True:
    n = int(input("Digite um número ou 0 para sair: "))
    if n == 0:
        break
    L.append(n)

    print("Elementos atuais da lista:")
    for elemento in L:
        print(elemento)

#B) A alternativa anterior teve êxito? Justifique:
#RESPOSTA: Sim. A substituição do While pelo For, além de tornar o código mais visual, também elimina a necessidade declarativa e de controle manual do índice.
#Mas devemos ter atenção, pois o While necessita de estrutura de dados condicional e o for requer uma sequência o intervalo previamente definido.
```
