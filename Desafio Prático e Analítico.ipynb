```

#1- Observe o código apresentado a seguir e veja outras formas de inserir dados em uma lista.

#A) Refaça o código utilizando a estrutura FOR:
numeros = [0, 0, 0, 0, 0]

for x in range(5):
    numeros[x] = float(input("Digite um número qualquer: "))

for y in range(5):
    print(f"{numeros[y]}")

#B) O que acontece se colocarmos while x <= 4? Justifique:
#RESPOSTA: Funcionalmente, terá o mesmo resultado, pois os índices de uma lista de tamanho 5 variam de 0 a 4. As condições x < 5 e x <= 4 executam exatamente para x = 0, 1, 2, 3, 4.

#C) Esse código funcionaria caso quiséssemos inserir mais de 5 elementos? Justifique:
#RESPOSTA: Não, pois como a lista foi inicializada com 5 elementos e o valor do while está fixado em 5. Caso tentarmos acessar um valor maior, acorrerá um erro, pois o índice estará fora do alcance.

#D) Qual a problemática da alternativa anterior (c) para a realidade dos códigos?
#RESPOSTA: Utilizar um tamanho fixo engessa o programa, impedindo que ele trate dados de tamanho dinâmico ou definido pelo usuário em tempo de execução.

```

```
#E) Refaça o código melhorando os problemas encontrados na alternativa (c e d):

numeros = []
qtd = int(input("Quantos números deseja inserir? "))

for i in range(qtd):
    num = float(input(f"Digite o {i+1}º número: "))
    numeros.append(num)

print("\nNúmeros digitados:")
for num in numeros:
    print(num)
```
